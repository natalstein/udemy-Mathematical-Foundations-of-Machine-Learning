Partial derivatives of the cylinder volume $V = \pi r^2 l$ describe how the volume changes with respect to each variable.

The derivative with respect to length is $\frac{\partial V}{\partial l} = \pi r^2$, representing how volume changes as the length varies.

The derivative with respect to radius is $\frac{\partial V}{\partial r} = 2 \pi r l$, representing how volume changes as the radius of the base varies.

In this exercise, we will compare autodiff results from PyTorch with those obtained using analytical differentiation and NumPy.

```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import sympy as sym
```

**Define functions**





```python

def cylinder_volume(my_r, my_l):
    return np.pi * my_r**2 * my_l
def partial_volume_r(my_r, my_l):
    return 2 * np.pi * my_r * my_l
def partial_volume_l(my_r, my_l):
    return np.pi * my_r**2
```

**Create variable to use in numpy and torch methods**

```python
r = 3.0
l = 5.0
r_t = torch.tensor(r, requires_grad=True)
l_t = torch.tensor(l, requires_grad=True)

```

**Compute and compare results**

```python
v = cylinder_volume(r,l)
v_t = cylinder_volume(r_t,l_t)
v, v_t

```

# **Autodifferentiate v_t**

```python

v_t.backward()
print(r_t.grad)
print(partial_volume_r(r,l))
```

**Exercise 1**

Find derivatives of $z = y^3 + 5xy$

Partial derivative of z with respect to y is $\frac{\partial z}{\partial y} = 3y^2 + 5x$

Partial derivative of z with respect to x is $\frac{\partial z}{\partial x} = 5y$



```python
#define all function
def z(my_y,my_x):
  return my_y**3 + 5*my_y*my_x
def partial_z_y(my_y,my_x):
  return 3*my_y**2 + 5*my_x
def partial_z_x(my_y,my_x):
  return 5*my_y
```

```python
#define variables in torch simple computation
x = 1.0
y = 1.0
x_t = torch.tensor(x, requires_grad=True)
y_t = torch.tensor(y, requires_grad=True)
```

```python
z_t = z(y_t,x_t)
z_t.backward()
print(y_t.grad, x_t.grad)
print(partial_z_y(y,x), partial_z_x(y,x))
```

Find all the derivatives of $a = 2\pi r^2 + 2\pi r h$.

Partial derivative of a with respect to r is $\frac{\partial a}{\partial r} = 4\pi r +2\pi h$

Partial derivative of a with respect to h is $\frac{\partial a}{\partial h} = 2\pi r$

```python
r_s = sym.symbols('r_s')
h_s = sym.symbols('h_s')
a = 2*sym.pi*r_s**2 + 2*sym.pi*h_s*r_s
a_r = sym.diff(a,r_s)
a_h = sym.diff(a,h_s)
a_r, a_h

```

```python
#define functions
def a(my_r, my_h):
  return 2*np.pi*my_r**2 + 2*np.pi*my_h
def partial_a_r(my_r, my_h):
  return 4*np.pi*my_r + 2*np.pi*my_h
def partial_a_h(my_r, my_h):
  return 2*np.pi * my_r
```

```python
#define variables
h =1.0
r = 1.0
h_t = torch.tensor(h, requires_grad=True)
r_t = torch.tensor(r, requires_grad=True)
```

```python
#compute and compare
a_t = a(r_t,h_t)
a_t.backward()
print(r_t.grad, partial_a_r(r,h))
print(h_t.grad, partial_a_h(r,h))

```
